## One Python script: from CAD to solution

* **Geometry**: OpenCascade CAD kernel, parametric shapes built in Python, STEP import
* **Mesh**: Netgen mesh generator, 2D / 3D / surfaces, curved high-order elements
* **Physics**: variational formulations in near-mathematical notation, compiled to fast C++ kernels
* **Solvers**: sparse direct, BDDC, AMG, multigrid; shared memory, MPI, and GPU (CUDA, Metal)
* **Visualization**: interactive WebGL viewer in Jupyter and in the browser

Open source (LGPL), `pip install ngsolve`, developed over more than 20 years at JKU Linz, RWTH Aachen and TU Wien.

## What makes it different

* **High order everywhere**: arbitrary polynomial order on curved meshes, for accurate fields with few unknowns
* **The right function spaces**: $H^1$, $H(\operatorname{curl})$, $H(\operatorname{div})$, $H(\operatorname{div} \operatorname{div})$, DG and hybrid DG, spaces on surfaces and facets
* **Multiphysics by composition**: product spaces, coupled and mixed formulations
* **Nonlinear problems from energies**: write down the energy, automatic differentiation provides residual and Jacobian for Newton
* **Matrix-free and GPU** operator application for large problems

## Simulation data at scale

* parametric geometries: every design is a function call, loop over thousands of variants
* runs headless on workstations, clusters and GPUs
* meshes and fields are available as NumPy arrays; solutions can be evaluated on arbitrary point clouds
* high-order accuracy and error control give clean, reliable data

## Showcases

| | Physics | NGSolve features |
|---|---|---|
| [Capacitor](wta/capacitor.ipynb) | 2D electrostatics | geometry to field in a few lines; corner singularities resolved by hp-refinement, rounded CAD edges and curved high-order elements |
| [Elasticity](wta/elasticity3D.ipynb) | 3D linear elasticity | CAD with chamfers, curved high-order tets, BDDC-preconditioned CG |
| [Coil](wta/coil.ipynb) | magnetostatics | two coupled problems, current flow plus $H(\operatorname{curl})$ Maxwell, field lines |
| [de Rham complex](wta/derham.ipynb) | mathematics | exact sequences and exact conservation laws; Piola transformation keeps fields divergence-free on deformed domains |
| [Tesla valve](wta/tesla.ipynb) | incompressible Navier–Stokes | exactly divergence-free $H(\operatorname{div})$ hybrid DG, matrix-free convection, IMEX time stepping |
| [Shells](wta/shell.ipynb), [Sydney](wta/Sydney.ipynb) | nonlinear shell mechanics | surface meshes, mixed TDNNS method, Newton with automatic differentiation; the Sydney Opera House roof |
